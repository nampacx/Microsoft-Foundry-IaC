```polyglot-notebook
#r "nuget: Azure.AI.Translation.Text, 2.0.0-beta.1"
#r "nuget: Azure.Identity"

using Azure.AI.Translation.Text;
using Azure.Identity;
using System.Text.Json;
```

```polyglot-notebook
var translationEndpoint = "https://kk-tt.cognitiveservices.azure.com/";

var foundryEndpoint = "https://kk-tt-foundry.cognitiveservices.azure.com/";

var targetLanguage = "de";
var textToTranslate = "Hello, how are you?";
```

```polyglot-notebook
var credential = new DefaultAzureCredential();
var  translationClient = new TextTranslationClient(credential, new Uri(translationEndpoint));

var translationResult = translationClient.Translate(targetLanguage, textToTranslate);

foreach (var translation in translationResult.Value)
{
    Console.WriteLine($"Translated Text: {translation.Translations[0].Text}");
}

var foundryClient = new TextTranslationClient(credential, new Uri(foundryEndpoint));
var foundryTranslationResult = foundryClient.Translate(targetLanguage, textToTranslate);


Assert.Equal(translationResult.Value[0].Translations[0].Text, foundryTranslationResult.Value[0].Translations[0].Text);
```
